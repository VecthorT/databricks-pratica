```python
from pyspark.sql.functions import current_timestamp, input_file_name, lit
```

```python
base_arquivos = '/Volumes/workspace/raw/kaggle'
df_cards = spark.read\
            .option('header', True)\
            .option('inferSchema', False)\
            .csv(f'{base_arquivos}/cards_data.csv')\
            .withColumn('_dt_ingestao', current_timestamp())\
df_cards.show(10)
```

```python
df_cards.write\
    .format('delta')\
    .mode('overwrite')\
    .saveAsTable(f'workspace.bronze.cards_data')
```
